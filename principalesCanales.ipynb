```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
#data = pd.read_csv('inmuebles2.csv')
data = pd.read_excel('Principales_Canales.xlsx')
data.head(5)
```

```python
#Identificar valores nulos por columna
valores_nulos=data.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data.isnull().sum().sum()
valores_nulos
```

```python
#Ver el número de filas en el archivo
data.shape[0]
```

```python
# Eliminar columna innecesarias

#La columna "Unnamed: 31" tiene 5 celdas NaN
#Ya que el archivo tiene una extensión de 58 filas
#es mejor eliminar toda la columna

data1 = data.drop(["Unnamed: 31"], axis=1)

data1.head()
```

```python
# Sustituir valores nulos por un string en concreto

#Al ser una columna de texto, solo hace falta sustituirlos por un texto especifico.
data1["Unnamed: 0"] = data1["Unnamed: 0"].fillna("Sin dato")
```

```python
# Sustituir valores nulos por un valor numérico en concreto

#Al ser números representativos y no calculados, los sustituimos por valores fijos.
data1["Unnamed: 1"] = data1["Unnamed: 1"].fillna(0)
```

```python
# Sustituir valores nulos de las columnas mensuales por 0

#En este caso, el 0 funcionaría como un valor "simbolico" que
#significaria o representaria la ausencia de información
#para dicha columna

data1.loc[:, "Unnamed: 1":"Unnamed: 30"] = data1.loc[:, "Unnamed: 1":"Unnamed: 30"].fillna(0)
```

```python
# Sustituir valores nulos de la columna Unnamed: 32 por un string

# Esta columna contiene datos de resumen y diferentes tipos de información
# por lo que "Sin dato" representa la ausencia de información sin alterar los valores existentes

data1["Unnamed: 32"] = data1["Unnamed: 32"].fillna("Sin dato")
```

```python
# Sustituir valores nulos de la columna Unnamed: 33 por un string especifico.

#Al ser una columna de texto, solo hace falta sustituirlos por un texto especifico

data1["Unnamed: 33"] = data1["Unnamed: 33"].fillna("Sin dato")
```

```python
# Sustituir valores nulos de la columna Unnamed: 34 por un string especifico.

#Al ser una columna de texto, solo hace falta sustituirlos por un texto especifico

data1["Unnamed: 34"] = data1["Unnamed: 34"].fillna("Sin dato")
```

```python
#Comprobamos cantidad de valores nulos luego de la limpieza.

valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

```python
#Generar CSV limpio

data1.to_csv("Principales_Canales_Limpio.csv", index=False)
```
