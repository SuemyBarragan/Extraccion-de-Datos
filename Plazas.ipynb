```python
# Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carga desde un archivo .excel sin indice
data = pd.read_excel('Plazas.xlsx', index_col=False)
data.head(7)
```

```python
data.info()
```

```python
# Nombre de columnas
data.columns.tolist()
```

```python
# Eliminar columnas de "Unnamed"
data1= data.drop(columns=["Unnamed: 16",'Unnamed: 17','Unnamed: 18','Unnamed: 19'])
```

```python
# Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

```python
# Identificar valores nulos por columna
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
# Realizamos una copia del dataframe
data_copy = data1.copy()
```

```python
# Eliminamos las filas que no contienen nada de información
data2 = data_copy.drop(data_copy.index[279:1048549])
data3 = data2.drop(195)
```

```python
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
# Eliminar columna "Teléfono" por no tener información 
data3= data3.drop(columns=['Telefono'])
```

```python
# Columna de "Fecha Origen"
# Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill")
# Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
data3["Fecha Origen"] = data3["Fecha Origen"].ffill()
data3["Fecha Origen"] = data3["Fecha Origen"].bfill() 
```

```python
# Columna "Nombre"
# Sustituir valores nulos por un valor en  concreto 
data3["Nombre"] = data3["Nombre"].fillna("Nombre Desconocido")
```

```python
# Columna "Producto"
# Sustituir valores nulos por un valor en  concreto 
data3["Producto"] = data3["Producto"].fillna("Producto Desconocido")
```

```python
# Columna "Asesor"
# Sustituir valores nulos por un valor en  concreto 
data3["Asesor"] = data3["Asesor"].fillna("Nombre de Asesor Desconocido")
```

```python
# Columna "Origen"
# Sustituir valores nulos por un valor en  concreto 
data3["Origen"] = data3["Origen"].fillna("Origen Desconocido")
```

```python
# Columna "Seguimiento"
# Sustituir valores nulos por un valor en  concreto 
data3["Seguimiento"] = data3["Seguimiento"].fillna("No hay información")
```

```python
# Columna "Cita efectiva"
# Sustituir valores nulos por un valor en  concreto 
data3['Cita efetiva '] = data3['Cita efetiva '].fillna("No hay información")
```

```python
# Columna "PDM"
# Sustituir valores nulos por un valor en  concreto 
data3['PDM'] = data3['PDM'].fillna("No hay información")
```

```python
# Columna "SDC"
# Sustituir valores nulos por un valor en  concreto 
data3['SDC'] = data3['SDC'].fillna("No hay información")
```

```python
# Columna "Estatuts"
# Sustituir valores nulos por un valor en  concreto 
data3['Estatus'] = data3['Estatus'].fillna("No hay información")
```

```python
# Columnas de "Preguntas"
# Creación de variable de preguntas
preguntas = [
  "¿Por que no hay seguimiento?",
  "¿Por que no ha visitado la agencia?",
  "¿Por que no hubo prueba de manejo ?",
  "¿Por que no hubo solicitud de credito?",
  "¿Por que no se ha formalizado la compra?"
]

# Sustituir valores nulos por un valor en concreto
for columna in preguntas:
    data3[columna] = data3[columna].fillna("No respondió")
```

```python
# Comprobación datos nulos 
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
# Convertir DataFrame a CSV
data3.to_csv("Plazas_Sin_Nulos.csv")
```
