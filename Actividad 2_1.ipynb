```python
# Realizamos una copia del dataframe
data1 = data.copy()
```

```python
# Columna "Potencial de compra"
# Sustituir valores nulos con mediana

# Convertirla a numérica
data1["Potencial de compra"] = pd.to_numeric(data1["Potencial de compra"], errors="coerce")

# Reemplazar valores 
data1["Potencial de compra"]=data1["Potencial de compra"].fillna(round(data1["Potencial de compra"].median(),1))
```

```python
# Columna de "Fecha"
# #Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill")
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
data1["Fecha"] =data1["Fecha"].ffill()
data1["Fecha"] =data1["Fecha"].bfill() 
```

```python
# Columna "Nombre Cliente"
# Sustituir valores nulos por un valor en  concreto 
data1["Nombre Cliente "] = data1["Nombre Cliente "].fillna("Nombre Desconocido")
```

```python
# Columna "Telefono"
# Sustituir valores nulos por un valor en concreto
data1["Telefono "] = data1["Telefono "].fillna("No registrado")
```

```python
# Columna "BDC"
# Sustituir valores nulos por un valor en concreto
data1["BDC "] = data1["BDC "].fillna("Nombre Desconocido")
```

```python
# Columna "Estatus de Lead"
# Sustituir valores nulos por un valor en concreto
data1["Estatus de Lead"] = data1["Estatus de Lead"].fillna("Estatus Desconocido")
```

```python
# Columna "PDM"
# Sustituir valores nulos por un valor en concreto
data1["PDM"] = data1["PDM"].fillna("Desconocido")
```

```python
# Columna "SDC"
# Sustituir valores nulos por un valor en concreto
data1["SDC"] = data1["SDC"].fillna("Desconocido")
```

```python
# Columna "Venta"
# Sustituir valores nulos por un valor en concreto
data1["Venta"] = data1["Venta"].fillna("Desconocido")
```

```python
# Columna "Asesor Asignado"
# Sustituir valores nulos por un valor en concreto
data1["Asesor Asignado"] = data1["Asesor Asignado"].fillna("No asignado")
```

```python
# Columnas de "Preguntas"
# Creación de variable de preguntas
preguntas = [
    "¿Por que no ha visitado la agencia?",
    "¿Por que no hay solicitud de credito?",
    "¿Por que no hubo prueba de manejo ?",
    "¿Por que no hubo proceso wow?",
    "¿Por que se asigno antes de vistar piso?"
]

# Sustituir valores nulos por un valor en concreto
for columna in preguntas:
    data1[columna] = data1[columna].fillna("No respondió")
```

```python
# Comprobación datos nulos 
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Convertir DataFrame a CSV
data1.to_csv("Citas_Digital_Filtrado_Limpio.csv")
```
