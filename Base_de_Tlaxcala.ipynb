```python
# Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carga desde un archivo .excel sin indice
data = pd.read_excel('Base_de_Tlaxcala.xlsx', index_col=False)
data.head(5)
```

```python
data.info()
```

```python
# Nombre de columnas
data.columns.tolist()
```

```python
# Identificar valores nulos por dataframe
valores_nulos=data.isnull().sum().sum()
valores_nulos
```

```python
# Identificar valores nulos por columna
valores_nulos=data.isnull().sum()
valores_nulos
```

```python
# Realizamos una copia del dataframe
data_copy = data.copy()
```

```python
# Eliminar columna de "Teléfono" y "Correo"
# No hay valores en las columnas de Telefono y Correo, por lo que se eliminan del dataframe
data1 = data_copy.drop(columns=['Telefono', 'Correo'])
```

```python
# Columna "Producto"
# Sustituir valores nulos por un valor en concreto
data1["Producto"] = data1["Producto"].fillna("No especificado")
```

```python
# Comprobación datos nulos 
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Convertir DataFrame a CSV
data1.to_csv("Base_de_Tlaxcala_Sin_Nulos.csv")
```
